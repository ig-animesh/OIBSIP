## 6. Missing Data Analysis and Treatment Strategy

Missing values are handled according to the meaning and data type of each column.

### Strategy

- **Age:** Median imputation is used because age is numerical and the median is less affected by extreme values.
- **Annual_Income:** Median imputation is used because income can be skewed and may contain extreme values.
- **Purchase_Amount:** Median imputation is used because purchase amounts can also be skewed by unusually large purchases.
- **Gender:** Mode imputation is used because gender is categorical and the most frequent valid category provides a reasonable replacement.
- **City:** Mode imputation is used because city is categorical.
- **Purchase_Date:** Rows with missing purchase dates will be removed because a transaction without a date cannot be reliably assigned to a specific time period.
- **Payment_Method:** Mode imputation is used because it is a categorical variable.
- **Customer_Name:** Rows where the customer name is completely missing will be removed because the record lacks an important identifying field.
- **Membership_Status:** No missing values are currently present, so no imputation is required.
- **Customer_ID:** No missing values are currently present. IDs should not be statistically imputed.

### Missing Value Report

```python
missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().mean() * 100).round(2)
})

missing_report = missing_report.sort_values(
    "Missing_Count",
    ascending=False
)

missing_report
```

### Preserving Missing Counts

```python
missing_before = df.isnull().sum().sum()
duplicates_before = df.duplicated().sum()
rows_before = len(df)

print("Total missing values:", missing_before)
print("Duplicate rows:", duplicates_before)
print("Rows:", rows_before)
```

### Row Deletion Decision

Rows with a missing Customer_Name and Purchase_Date are removed because they lack sufficient identifying and transaction information.

Statistical imputation is inappropriate for these fields because creating a fictional customer identity or transaction date could introduce incorrect information into the dataset.

```python
rows_to_delete = df["Customer_Name"].isna() & df["Purchase_Date"].isna()

print("Rows identified for deletion:", rows_to_delete.sum())

df = df[~rows_to_delete].copy()

print("Rows after deletion:", len(df))
```

## 7. Data Type Correction

Annual income and purchase amount contain values with currency symbols, spaces, and numeric values stored as strings.

Currency symbols and unnecessary whitespace are removed before converting the columns to numeric values.

```python
df["Annual_Income"] = (
    df["Annual_Income"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["Purchase_Amount"] = (
    df["Purchase_Amount"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["Annual_Income"] = pd.to_numeric(df["Annual_Income"], errors="coerce")
df["Purchase_Amount"] = pd.to_numeric(df["Purchase_Amount"], errors="coerce")
```

### Date Standardisation

Purchase_Date contains multiple date formats such as:

- YYYY-MM-DD
- DD/MM/YYYY
- YYYY/MM/DD
- DD-MM-YYYY
- Month DD, YYYY

The values are converted into a single pandas datetime format. Invalid or missing dates are converted to NaT and will be handled appropriately.

```python
df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"], format="%Y-%m-%d")
```

### Missing Data Rows

```python
missing_date_rows = df["Purchase_Date"].isna().sum()

print("Rows with missing/invalid Purchase_Date:", missing_date_rows)
```

```python
df = df.dropna(subset=["Purchase_Date"]).copy()

print("Rows after removing missing dates:", len(df))
```

## 8. Categorical Data Standardisation

Categorical values are standardised so that the same category is represented consistently.

For Gender:

- Male
- male
- M

are converted to:

**Male**

Similarly:

- Female
- female
- FEMALE
- F

are converted to:

**Female**

Invalid values such as "Unknown" are treated as missing and subsequently replaced with the mode.

```python
df["Gender"] = (
    df["Gender"]
    .astype("string")
    .str.strip()
    .str.lower()
)

gender_mapping = {
    "male": "Male",
    "m": "Male",
    "female": "Female",
    "f": "Female"
}

df["Gender"] = df["Gender"].map(gender_mapping)
```

### Standardise City

```python
df["City"] = (
    df["City"]
    .astype("string")
    .str.strip()
    .str.title()
)

valid_cities = [
    "Kanpur",
    "Delhi",
    "Lucknow",
    "Noida",
    "Jaipur",
    "Agra"
]

df.loc[~df["City"].isin(valid_cities), "City"] = pd.NA
```

### Membership Status Standardisation

Membership values are standardised into two valid categories:

- Yes
- No

Values such as "yes", "NO", and "Y" are converted to the appropriate standard representation.

Invalid values such as "Maybe" are treated as missing because they do not represent a valid membership category.

```python
df["Membership_Status"] = (
    df["Membership_Status"]
    .astype("string")
    .str.strip()
    .str.lower()
)

membership_mapping = {
    "yes": "Yes",
    "y": "Yes",
    "no": "No",
    "n": "No"
}

df["Membership_Status"] = df["Membership_Status"].map(membership_mapping)
```

### Payment Method

```python
df["Payment_Method"] = (
    df["Payment_Method"]
    .astype("string")
    .str.strip()
    .str.lower()
)

payment_mapping = {
    "cash": "Cash",
    "upi": "UPI",
    "credit card": "Credit Card",
    "debit card": "Debit Card"
}

df["Payment_Method"] = df["Payment_Method"].map(payment_mapping)
```

### Missing Value Imputation

```python
# Numerical columns – median imputation
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Annual_Income"] = df["Annual_Income"].fillna(
    df["Annual_Income"].median()
)

df["Purchase_Amount"] = df["Purchase_Amount"].fillna(
    df["Purchase_Amount"].median()
)

# Categorical columns – mode imputation
categorical_columns = [
    "Gender",
    "City",
    "Membership_Status",
    "Payment_Method"
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after imputation:")
print(df.isnull().sum())
```

## 9. Duplicate Detection and Removal

Exact duplicate rows can cause double-counting and distort statistical analysis.

Duplicate rows were identified using pandas' `duplicated()` method.

Because these records contain exactly the same information as another record, retaining them would not add analytical value. Therefore, exact duplicate rows are removed.

```python
duplicates_found = df.duplicated().sum()

print("Duplicate rows before removal:", duplicates_found)

df = df.drop_duplicates().copy()

print("Duplicate rows after removal:", df.duplicated().sum())
print("Rows after duplicate removal:", len(df))
```

## 10. Outlier Detection Using the IQR Method

The Interquartile Range (IQR) method is used to identify potential outliers in numerical columns.

### Formula

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values below the lower bound or above the upper bound are considered potential outliers.

Outliers are not automatically errors. Each numerical column is evaluated before deciding whether to remove, cap, or retain the observations.

### IQR Function

```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]
    
    return Q1, Q3, IQR, lower_bound, upper_bound, outliers
```

### Outlier Report

```python
outlier_report = []

numeric_columns = [
    "Age",
    "Annual_Income",
    "Purchase_Amount"
]

for col in numeric_columns:
    Q1, Q3, IQR, lower, upper, outliers = detect_outliers_iqr(df, col)
    
    outlier_report.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_report)

outlier_report
```

### Inspecting Outliers

```python
for col in numeric_columns:
    Q1, Q3, IQR, lower, upper, outliers = detect_outliers_iqr(df, col)
    
    print(f"\n{'='*50}")
    print(f"Column: {col}")
    print(f"Lower Bound: {lower:.2f}")
    print(f"Upper Bound: {upper:.2f}")
    print(f"Outlier Count: {len(outliers)}")
    
    if len(outliers) > 0:
        display(outliers[["Customer_ID", col]])
```

### Logical Range Validation

Statistical outlier detection alone is not sufficient.

Some values can be identified as invalid based on domain logic:

- Customer age should be between 18 and 100.
- Annual income should be positive.
- Purchase amount should be greater than or equal to zero.

These logical rules are therefore applied in addition to IQR-based detection.

```python
invalid_age = (df["Age"] < 18) | (df["Age"] > 100)
invalid_income = df["Annual_Income"] <= 0
invalid_purchase = df["Purchase_Amount"] < 0

print("Invalid age records:", invalid_age.sum())
print("Invalid income records:", invalid_income.sum())
print("Invalid purchase amount records:", invalid_purchase.sum())
```

## 11. Outlier Treatment Decision

Outliers are treated according to whether they represent possible data-entry errors or legitimate extreme observations.

### Age

Age values below 18 or above 100 are considered invalid for this customer dataset. These values are replaced with the median age because the extreme values are clearly inconsistent with the intended domain.

### Annual Income

Extremely high income values may represent legitimate high-income customers. However, the intentionally inserted extreme value is considered a data-entry anomaly in this exercise. It will be capped at the IQR upper bound rather than deleting the entire customer record.

### Purchase Amount

Negative purchase amounts are invalid because a purchase amount cannot be negative in this dataset. These values are replaced with the median purchase amount.

Extremely high but positive purchase amounts are treated as potential legitimate transactions and are retained unless they violate the defined business rules.

### Rationale

Capping is preferred for extreme income values because it preserves the customer record while reducing the influence of an extreme value. Invalid values are corrected using median-based treatment rather than unnecessarily deleting otherwise useful records.

### Fixing Invalid Ages

```python
age_median = df["Age"].median()

df.loc[
    (df["Age"] < 18) | (df["Age"] > 100),
    "Age"
] = age_median
```

### Fixing Negative Purchase Amounts

```python
purchase_median = df["Purchase_Amount"].median()

df.loc[
    df["Purchase_Amount"] < 0,
    "Purchase_Amount"
] = purchase_median
```

### Cap Extreme Income

```python
Q1 = df["Annual_Income"].quantile(0.25)
Q3 = df["Annual_Income"].quantile(0.75)

IQR = Q3 - Q1

income_upper_bound = Q3 + 1.5 * IQR

df["Annual_Income"] = df["Annual_Income"].clip(
    upper=income_upper_bound
)

print("Income upper cap:", income_upper_bound)
```

## 12. Final Data Type Correction & Verification

After cleaning, each column is explicitly converted to its intended data type.

This ensures that identifiers remain strings, numerical measures are numeric, and purchase dates are stored as pandas datetime values.

```python
df["Customer_ID"] = df["Customer_ID"].astype("string")
df["Customer_Name"] = df["Customer_Name"].astype("string")
df["Gender"] = df["Gender"].astype("string")
df["Age"] = df["Age"].round().astype("int64")
df["City"] = df["City"].astype("string")
df["Annual_Income"] = df["Annual_Income"].astype("float64")
df["Purchase_Amount"] = df["Purchase_Amount"].astype("float64")
df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"])
df["Membership_Status"] = df["Membership_Status"].astype("string")
df["Payment_Method"] = df["Payment_Method"].astype("string")
```

- Number of columns
